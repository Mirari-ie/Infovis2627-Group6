```python
categorical = set(df.select_dtypes(include=["object", "str"]).columns)

units = {
    "age": "years",
    "avg_daily_sm_hours": "hours/day",
    "years_on_social_media": "years",
    "avg_daily_screen_time_hours": "hours/day",
    "avg_sleep_hours": "hours/night",
    "avg_work_study_hours_per_day": "hours/day",
    "dopamine_rush_feel_score": "1–10 score",
    "reward_seeking_behavior": "1–10 score",
    "fomo_score": "1–10 score",
    "comparison_to_others_score": "1–10 score",
    "validation_seeking_score": "1–10 score",
    "attention_span_minutes": "minutes",
    "deep_work_duration_minutes": "minutes",
    "productivity_self_rating": "1–10 score",
    "tasks_completed_per_day": "tasks/day",
    "work_quality_self_rating": "1–10 score",
    "reading_duration_minutes_per_day": "minutes/day",
    "decision_fatigue_score": "1–10 score",
    "anxiety_when_phone_unavailable": "1–10 score",
    "restlessness_score": "1–10 score",
    "withdrawal_attempt_count": "attempts",
    "longest_detox_days": "days",
    "detox_relapse_reason": "coded numeric score; unit undocumented",
    "dopamine_dysregulation_index": "composite index (0–10)",
    "digital_wellbeing_score": "composite index (0–10)",
    "sm_addiction_risk_score": "composite index (0–10)",
    "notes": "numeric value; unit undocumented",
}

special_desc = {
    "participant_id": "Unique participant identifier; nominal identifier.",
    "data_source": "Source label for the records.",
    "synthetic_flag": "Binary flag indicating whether the data are synthetic.",
    "age": "Participant age.",
    "age_group": "Age band derived from age.",
    "gender": "Self-reported gender category.",
    "race_ethnicity": "Race/ethnicity category.",
    "nationality": "Nationality category; overlaps strongly with country_of_residence.",
    "country_of_residence": "Country where the participant resides; geographic categorical field.",
    "education_level": "Education category.",
    "employment_status": "Employment status category.",
    "occupation_category": "Broad occupation category.",
    "relationship_status": "Relationship status category.",
    "primary_platform": "Main social-media platform.",
    "secondary_platform": "Secondary social-media platform, if any.",
    "productivity_decline_score": "Ordinal outcome describing reported productivity decline.",
}

def field_description(c):
    return special_desc.get(
        c,
        "Categorical behavioral/self-report variable." if c in categorical
        else "Numeric behavioral, psychological or productivity measure."
    )

def temporal_geo(c):
    return "Geographic" if c in {"country_of_residence", "nationality"} else "No temporal dimension"

def domain(c):
    s = df[c]
    if c in categorical:
        n = s.dropna().nunique()
        if n <= 12:
            return ", ".join(map(str, sorted(s.dropna().unique())))
        return f"{n} distinct categories"
    return f"{s.min():g} to {s.max():g}"

metadata_rows = []
for c in df.columns:
    metadata_rows.append({
        "Field": c,
        "Data type": str(df[c].dtype),
        "Kind": "Categorical" if c in categorical else "Numerical",
        "Temporal / geographic": temporal_geo(c),
        "Domain / range": domain(c),
        "Unit / precision": units.get(c, "Category label" if c in categorical else "Numeric value; precision as stored"),
        "Missing": int(df[c].isna().sum()),
        "Unique": int(df[c].nunique(dropna=True)),
        "Description": field_description(c),
    })

field_dictionary = pd.DataFrame(metadata_rows)
```

```python
from IPython.display import display, HTML

html_table = field_dictionary.to_html(
    index=False,
    classes="dataframe",
    escape=False
)

display(HTML(f"""
<div style="
    max-height: 500px;
    overflow: auto;
    border: 1px solid #ddd;
">
    {html_table}
</div>
"""))
```
